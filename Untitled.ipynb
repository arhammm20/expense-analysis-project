```python
import pandas as pd
```

```python
df=pd.read_csv("messy_expense_dataset.csv")

df.isnull().sum()
df.shape
df.info()
df.describe()
df.head(50)
```

```python
df=df.drop_duplicates()
print(f'rows after removing duplicates: {df.shape}')
```

```python
df['Date']=pd.to_datetime(df['Date'],errors='coerce')
df['Date'].isnull().sum()
df=df.dropna(subset=['Date'])
```

```python

df['Amount']=df['Amount'].astype(str)
df['Amount']=df['Amount'].str.replace('₹', '', regex=False)
df['Amount']=pd.to_numeric(df['Amount'],errors='coerce')
df=df.dropna(subset=['Amount'])
```

```python
df['Type']=df['Type'].astype(str).str.lower()
df['Type'] = df['Type'].replace({
    'debit': 'Debit',
    'credit': 'Credit'
})
df = df[df['Type'].isin(['Debit', 'Credit'])]
```

```python
df.info()
df.isnull().sum()
df.head()
```

```python
df.to_csv("cleaned_expense_data.csv", index=False)         

'''  errors='coerce' ke baad null drop bhoolna
 Amount clean kiye bina analysis start kar dena
 Type column inconsistent chhod dena '''
```

```python
df['Month']=df['Date'].dt.month
df['Day']=df['Date'].dt.day
df['WeekDay']=df['Date'].dt.day_name()
df.head()
```

```python
'''df['Transaction'] = df['Type'].map({
    'Debit': 'Expense',
    'Credit': 'Income'
})'''
def expense(txt):
    if txt=='Debit':
        return 'Expense'
    else:
        return 'Income'
df['Transaction']=df['Type'].apply(expense)
df.head()
```

```python
df['Description'].value_counts()
```

```python
df['Description']=df['Description'].str.lower()
df['Description'].value_counts()
def categ(desc):
    if "swiggy" in desc or "zomato" in desc:
        return "food"
    elif "flipkart" in desc or "amazon" in desc:
        return "shopping"
    elif "electricity" in desc or "rent" in desc:
        return "bills"
    elif "uber" in desc:
        return "transport"
    elif "gym" in desc:
        return "health"
    else:
        return "other"
df["Category"]=df['Description'].apply(categ)        
df['Category'].value_counts()
df[['Description','Category']].sample(10)
```
